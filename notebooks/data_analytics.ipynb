```python
import sqlite3
import pandas as pd
conn = sqlite3.connect(r'C:\Users\nihad\OneDrive\Desktop\project 3.1\project 3.1\data\raw\ipl.db')


def q(sql):
    return pd.read_sql_query(sql, conn)
```

```python
def run_sql_file(path): 
    with open(path, 'r') as f:
        script = f.read()
    conn.executescript(script)
    conn.commit()    
    print('ran',path)
```

```python
q("""
    -- what is the strike rate of every batter in the death overs

    select batter,round(100.0*sum(batter_runs)/sum(
      CASE
        when is_wide_ball=0 AND is_no_ball=0
        then 1
        else 0
      END ),22)as strike_rate

from deliveries
where over_number>=15 AND is_super_over=0
group by batter
ORDER BY strike_rate desc;
    """)
```
